```python
df_cleaned = pd.read_csv('df_cleaned.csv')
```

```python
# Change to let the user select their stocks later
selected_stocks = ['AAPL', 'MSFT', 'GOOG']

df_stocks = df_cleaned[selected_stocks]

returns = df_stocks.pct_change().dropna()

# Rough estimate of yearly returns
avg_yr_returns = returns.mean() * 252 #approximate trading days in a year
```

```python
# Calculate correlation between the daily stock returns
stock_corr = returns.corr()
```
